```python
l=[87,98,54,123,987,323]
largest = l[0]
index = 0
for i in range(len(l)):
  if largest < l[i]:
    largest = l[i]
    index = i
print(largest)
print(index)
```

```python
l=[87,98,54,123,987,323]
largest = secondlargest = l[0]
for i in range(len(l)):
  if largest < l[i]:
    secondlargest = largest
    largest = l[i]
  elif secondlargest < l[i]:
    secondlargest = l[i]
print(secondlargest)
```

```python
a = [12,19,14,15,16]
for i in range(len(a)-1):
  if a[i] < a[i+1]:
    continue
  else:
    print("your list is not sorted")
    break
else:
  print("your list is sorted")
```
