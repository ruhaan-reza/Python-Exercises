**Write a function to count the frequency of elements in a list.**

```python
def count_freq(l):
  freq = {}
  for item in l:
    if item in freq:
      freq[item] += 1
    else:
      freq[item] = 1
  return freq
l=[1,2,3,3,3,4,5,6,7]
count_freq(l)
```

**Write a function that returns the factorial of a number.**

```python
def fact(n):
  fact = 1
  for i in range(1,n+1):
    fact = fact*i
  return fact
print(fact(5))
```
