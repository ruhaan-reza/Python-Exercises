```python
def pallindrome(s):
  b = ""
  for ch in s:
    b = ch + b
  if b == s:
    print("it is pallindrome")
  else:
    print("it is not pallindrome")
```

```python
pallindrome("naman")
```

```python
def pallindrome(st):
  rev = ""
  for i in range(len(st)-1,-1,-1):
    rev = rev + st[i]
  if rev == st:
    print(f"{st} is pallindrome")
  else:
    print(f"{st} is not pallindrome")
```

```python
pallindrome("ruhaan")
```
